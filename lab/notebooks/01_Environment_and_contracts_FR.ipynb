## 2. Un paquet est visible depuis un interpréteur

Prédire si NumPy est disponible ici. Cette inspection ne fait aucune installation. Une absence de `python-dotenv` ne bloque pas les calculs Python standard. `.venv` est un environnement Python ; `.env` est un fichier de configuration qui nécessite un chargement explicite, détaillé dans `../ENVIRONMENT.md`.

````python
import importlib.util

for module_name, package_name in (("numpy", "numpy"), ("ipykernel", "ipykernel"), ("dotenv", "python-dotenv")):
    available = importlib.util.find_spec(module_name) is not None
    print(package_name, ":", "présent" if available else "absent")
````

## 3. Afficher et renvoyer

Avant exécution : combien de lignes seront affichées ? Quelle sera la valeur de `reported` ? Quelle variable pourra entrer dans une addition ?

````python
def print_count(values: list) -> None:
    print(f"Count: {len(values)}")
#  pas de return

def count_items(values: list) -> int:
    return len(values)


records = ["A", "B", "C"]
reported = print_count(records)
count = count_items(records)
print("reported :", repr(reported), type(reported).__name__)
print("count :", repr(count), type(count).__name__)
print("count + 2 :", count + 2)
````

## 4. Une annotation ne transforme pas le résultat

Le contrat écrit promet un nombre entier. Prédire le type réellement renvoyé. Observer `repr()` pour distinguer une chaîne de caractères d’une valeur numérique.

````python
def summarize_count(values: list) -> int:
    # Intentional defect: the implementation violates its numeric contract.
    return f"{len(values)} records"


result = summarize_count(records)
print("Valeur :", repr(result))
print("Type :", type(result).__name__)
print("Respecte le type attendu ?", isinstance(result, int))
````

## 5. Une preuve réutilisable

Ces assertions contrôlent la fonction numérique `count_items`. Elles réussissent aussi dans un noyau neuf. Quelles erreurs détecteraient-elles ? Quelles erreurs ne couvrent-elles pas ?

````python
assert count_items([]) == 0
assert count_items(["A", "B", "C"]) == 3
assert isinstance(count_items(["A"]), int)
print("Trois contrôles manuels réussis pour count_items.")
````
