## Option A — Deux fonctions et des valeurs manquantes

Lire les slides 5–8 du deck de l'après-midi. Compléter les deux fonctions.
`known_minutes` retire seulement les valeurs manquantes.
`fraction_at_or_above` réutilise la première fonction et inclut le seuil.
Zéro est une valeur connue.

````python
values = pd.Series(
    [0, 120, pd.NA, 180],
    dtype="Float64"
)

print(values)
# values.describe()
````

````python
def known_minutes(values: pd.Series) -> pd.Series:
    """Return the observed minute values."""
    return values.dropna()

print(known_minutes(values))

def fraction_at_or_above(values: pd.Series, threshold: float) -> float:
    """Return the fraction of observed values at or above threshold."""
    known = known_minutes(values)

    if known.empty:
        return 0.0

    return float((known >= threshold).sum()) / len(known)

print(fraction_at_or_above(values, 120))
````

````python
option_a_ready = False
try:
    clean = known_minutes(values)
    fraction = fraction_at_or_above(values, 180)
    assert isinstance(clean, pd.Series)
    assert clean.tolist() == [0, 120, 180]
    assert clean.mean() == 100
    assert np.isclose(fraction, 1 / 3)
    option_a_ready = True
    print("Option A validée.")
except Exception as error:
    print(f"Option A à compléter: {type(error).__name__}: {error}")
````

````python
class TemperatureSeries:
    """Store Fahrenheit values and convert them to Celsius."""

    def __init__(self, fahrenheit: np.ndarray):
        self.fahrenheit = np.asarray(fahrenheit, dtype=float)

    def celsius(self) -> np.ndarray:
        return(self.fahrenheit - 32) * 5 / 9

test = TemperatureSeries(np.array([32, 68, 86]))
print(test.fahrenheit)
print(test.celsius())
````

## Option D — Boucle Python ou NumPy

Lire les slides 17–19. Compléter deux fonctions qui divisent les mêmes
valeurs par 60. Vérifier leur égalité **avant** la mesure. Les données sont
créées hors du chronomètre ; chaque mesure inclut le calcul et la création
du résultat.

````python
values = np.arange(300_000, dtype=float)

def loop_hours(values: list[float]) -> list[float]:
    """Convert minute values with an explicit Python loop."""
    result = []
    for value in values:
        result.append(value/60)
    return result

display(loop_hours(values.tolist()[:7]))

def numpy_hours(values: np.ndarray) -> np.ndarray:
    """Convert all values with one NumPy expression."""
    return  values / 60.0

display(numpy_hours(values[:7]).tolist())
````

````python
option_d_ready = False
values_array = np.arange(300_000, dtype=float)
values_list = values_array.tolist()

try:
    loop_result = loop_hours(values_list)
    numpy_result = numpy_hours(values_array)
    assert len(loop_result) == len(values_array)
    assert np.asarray(numpy_result).shape == values_array.shape
    assert np.allclose(loop_result, numpy_result)

    loop_times = timeit.repeat(
        lambda: loop_hours(values_list), repeat=5, number=1
    )
    numpy_times = timeit.repeat(
        lambda: numpy_hours(values_array), repeat=5, number=1
    )
    loop_median = float(np.median(loop_times))
    numpy_median = float(np.median(numpy_times))
    option_d_ready = True
    print(f"Boucle Python — médiane: {loop_median * 1000:.3f} ms")
    print(f"NumPy — médiane: {numpy_median * 1000:.3f} ms")
    print("Option D validée. Limitez votre conclusion à cette machine, cette taille et ce calcul.")
except Exception as error:
    print(f"Option D à compléter: {type(error).__name__}: {error}")
````
